# Regression (fitting the line using autodiff)
Also called as Autograd, Computational diff, Reversemode diff and Algorithmic diff. Better suited for functions with many inputs(like in the cas eof ML) and also for higher order derivatives


```python
# Auto diff with pytorch
import torch as pt
# import tensorflow as tf
# import numpy as np
import matplotlib.pyplot as plt
```

```python
x = pt.tensor([float(i) for i in list(range(8))])
x
```

```python
y = -0.5*x + 2 + pt.normal(mean=pt.zeros(8), std=0.2) # y here indicates the forgetful score for dosage of a drug x
y
```

```python
fig, ax = plt.subplots()
plt.title("Clinical trial")
plt.xlabel("Drug dosage")
plt.ylabel("Forgetfulness")
_ = ax.scatter(x,y)
```

```python
m = pt.tensor([0.9], requires_grad=True)
m
```

```python
b = pt.tensor([0.1], requires_grad=True)
b
```

```python
def regression(my_x, my_m, my_b):
    return my_m*my_x + my_b
```

```python
def regression_plot(my_x, my_y, my_m, my_b):
    
    fig, ax = plt.subplots()

    ax.scatter(my_x, my_y)
    
    x_min, x_max = ax.get_xlim()
    y_min = regression(x_min, my_m, my_b).detach().item()
    y_max = regression(x_max, my_m, my_b).detach().item()
    
    ax.set_xlim([x_min, x_max])
    _ = ax.plot([x_min, x_max], [y_min, y_max])

regression_plot(x, y, m, b)
```

```python
yhat = m*x + b
yhat
```

```python
def mse(my_yhat, my_y): 
    sigma = pt.sum((my_yhat - my_y)**2)
    return sigma/len(my_y)
```

```python
cost = mse(yhat, y)
cost
```

```python
cost.backward()
```

```python
m.grad
```

```python
b.grad
```

```python
optimizer = pt.optim.SGD([m, b], lr=0.01)
optimizer.step()
```

```python
m
```

```python
b
```

```python
C = mse(regression(x, m, b), y)
C
```

```python
epochs = 1000
for epoch in range(epochs):
    
    optimizer.zero_grad() # Reset gradients to zero; else they accumulate
    
    yhat = regression(x, m, b) # Step 1
    C = mse(yhat, y) # Step 2
    
    C.backward() # Step 3
    optimizer.step() # Step 4
    
    print('Epoch {}, cost {}, m grad {}, b grad {}'.format(epoch, '%.3g' % C.item(), '%.3g' % m.grad.item(), '%.3g' % b.grad.item()))
```

```python
regression_plot(x, y, m, b)
```
