# Regression (fitting the line using autodiff)
Also called as Autograd, Computational diff, Reversemode diff and Algorithmic diff. Better suited for functions with many inputs(like in the cas eof ML) and also for higher order derivatives


```python
# Auto diff with pytorch
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = tf.Variable([float(i) for i in list(range(8))])
x
```

```python
y = -0.5*x + 2 + tf.random.normal(shape=tf.shape(x), mean=0.0, stddev=0.2, dtype=tf.float32) 
y
```

```python
fig, ax = plt.subplots()
plt.title("Clinical trial")
plt.xlabel("Drug dosage")
plt.ylabel("Forgetfulness")
_ = ax.scatter(x,y)
```

```python
# Just initializing with starting values which are later continuosly adjusted through learning process
m = tf.Variable(0.9)
b = tf.Variable(0.2)
```

```python
def mse(my_yhat, my_y): 
    sigma = tf.reduce_sum((my_yhat - my_y)**2)
    return sigma/len(my_y)
```

```python
yhat = m*x + b
yhat
```

```python
def regression(my_x, my_m, my_b):
    return my_m*my_x + my_b
```

```python
def regression_plot(my_x, my_y, my_m, my_b):
    
    fig, ax = plt.subplots()

    ax.scatter(my_x, my_y)
    
    x_min, x_max = ax.get_xlim()
    y_min = regression(x_min, my_m, my_b).numpy()
    y_max = regression(x_max, my_m, my_b).numpy()
    
    ax.set_xlim([x_min, x_max])
    _ = ax.plot([x_min, x_max], [y_min, y_max])

regression_plot(x, y, m, b)
```

```python
with tf.GradientTape() as t:
    yhat = m*x + b
    cost = mse(yhat, y)

[dl_dm, dl_db] = t.gradient(cost, [m,b])

print(dl_dm)
print(dl_db)

# TODO perform linear regression with tensor flow
# def cost():
#     return mse(yhat, y)
# lr = 0.01
# optimizer = tf.compat.v1.train.GradientDescentOptimizer(lr)

# optimizer.minimize(loss, [m,b])
```

```python
def cost():
    return mse(yhat, y)
lr = 0.01
optimizer = tf.compat.v1.train.GradientDescentOptimizer(lr)
train  = optimizer.minimize(cost)
```

```python
# Global Variables Initializer
init = tf.compat.v1.global_variables_initializer()

print(tf.version.VERSION)
```

```python
training_epochs = 100
# Starting the Tensorflow Session
with tf.compat.v1.Session() as session:
     
    # Initializing the Variables
    session.run(init)
     
    # Iterating through all the epochs
    for epoch in range(training_epochs):
         
        # Feeding each data point into the optimizer using Feed Dictionary
        for (_x, _y) in zip(x, y):
            session.run(optimizer, feed_dict = {x : _x, y : _y})
         
        # Displaying the result after every 50 epochs
        # if (epoch + 1) % 50 == 0:
        #     # Calculating the cost a every epoch
        c = session.run(cost, feed_dict = {x : x, y : y})
        print("Epoch", (epoch + 1), ": cost =", c, "W =", session.run(m), "b =", session.run(b))
     
    # Storing necessary values to be used outside the Session
    training_cost = session.run(cost, feed_dict ={x: x, y: y})
    weight = session.run(m)
    bias = session.run(b)
```
